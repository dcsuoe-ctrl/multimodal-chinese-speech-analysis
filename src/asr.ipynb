```python
from transformers import pipeline
```

```python
def transcribe(audio_path):
    asr = pipeline(
        "automatic-speech-recognition",
        model="openai/whisper-small",
        return_timestamps=True
    )

    result = asr(
        audio_path,
        generate_kwargs={
            "language": "chinese",
            "task": "transcribe"
        }
    )

    return result
```
