```python
import parselmouth
```

```python
def extract_prosody(audio_path):
    sound = parselmouth.Sound(audio_path)

    pitch = sound.to_pitch()
    intensity = sound.to_intensity()

    return {
        "duration": sound.duration,
        "f0": pitch.selected_array["frequency"],
        "intensity": intensity.values[0]
    }
```
